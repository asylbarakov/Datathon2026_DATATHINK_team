```python
import pandas as pd
```

```python
df_2021 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/2t/2021/2_t_oplata_truda_2021.csv')
df_2022 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/2t/2022/2_t_oplata_truda_2022.csv')
df_2023 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/2t/2023/2_t_oplata_truda_2023.csv')
df_2024 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/2t/2024/sr_2t_ot_god_04_2024_parol_2025.csv')
```

```python
import gc

dfs_to_concat = [df_2021, df_2022, df_2023, df_2024]

df_combined = pd.concat(dfs_to_concat, ignore_index=True, copy=False)

del df_2021, df_2022, df_2023, df_2024
gc.collect()
```

```python
df_combined.to_csv('combined_data_2t.csv', index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Извлечем год из ID (формат S-2t-YYYY-XXXXXX)
df_combined['extracted_year'] = df_combined['ID'].str.split('-').str[2]

print("Распределение записей по годам (из ID):")
display(df_combined['extracted_year'].value_counts(dropna=False))
```

```python
print(f"Количество уникальных регионов (KATO): {df_combined['KATO'].nunique()}")
print(f"Количество уникальных отраслей (OKED): {df_combined['OKED'].nunique()}")

# Топ-10 регионов по количеству записей
plt.figure(figsize=(12, 5))
sns.countplot(data=df_combined, x='KATO', order=df_combined['KATO'].value_counts().index[:])
plt.title('Топ-15 регионов (KATO) по количеству наблюдений')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
