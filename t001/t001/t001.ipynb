```python
import pandas as pd
```

```python
df_2021 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/t001/2021/2021.csv')
df_2022 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/t001/2022/baza.csv')
df_2023 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/t001/2023/baza.csv')
df_2024 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/t001/2024/baza.csv')
```

```python
cols = list(df_2021.columns) + list(df_2022.columns) + list(df_2023.columns) + list(df_2024.columns)
```

```python
list(set(cols))
```

```python
import gc

dfs_to_concat = [df_2021, df_2022, df_2023, df_2024]

df_combined = pd.concat(dfs_to_concat, ignore_index=True, copy=False)

gc.collect()
```

```python
df_combined.to_csv('combined_data_t001.csv', index=False)
```
