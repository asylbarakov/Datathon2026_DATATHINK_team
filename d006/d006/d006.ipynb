```python
import pandas as pd
```

```python
df_2021 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/d006/2021/rio.csv')
df_2022 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/d006/2022/rio2022.csv')
df_2023 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/d006/2023/rio2023.csv')
df_2024 = pd.read_csv('/content/drive/MyDrive/Datathon2026/synthetic_microdata_2021-2024_20260921/d006/2024/rio_2024.csv')
```

```python
import gc

dfs_to_concat = [df_2021, df_2022, df_2023, df_2024]

df_combined = pd.concat(dfs_to_concat, ignore_index=True, copy=False)

gc.collect()
```

```python
df_combined.to_csv('combined_data_d006.csv', index=False)
```
